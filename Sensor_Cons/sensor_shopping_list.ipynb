## Group A · App 21 Motion Logger (also App 46 Digital Level) — Accelerometer

- **Sensor:** **MMA8452Q** (driver: `ACCEL`)
- **Buy:** [GY-45 MMA8452Q module](https://www.hotmcu.com/triple-axis-accelerometer-mma8452q-p-170.html) — ~$2.80 ·
  [ProtoCentral MMA8452Q (Qwiic)](https://protocentral.com/product/sparkfun-triple-axis-accelerometer-breakout-mma8452q-qwiic/) — ~$15
- **Driver:** `ACCEL` targets the **MMA8452Q** exactly (`ACCEL_DEVICE_ID 0x2A`, `WHO_AM_I` checked on init).
- **Wiring:** I²C, address 0x1C (SA0 low) / 0x1D (SA0 high).
- **Watch out:** Adafruit #2019 is the **MMA8451** (WHO_AM_I 0x1A) — not the part this driver checks for.
- **On-board note:** App 21 also uses the on-board L3GD20 gyro (`MEMS`) for rotation; only the accelerometer needs buying.

## Group A · App 22 Temperature & Humidity (also 43, 45, 50) — DHT22

- **Sensor:** **DHT22** / DHT11 (driver: `DHT`)
- **Buy:** [Waveshare DHT22 module](https://www.waveshare.com/dht22-temperature-humidity-sensor.htm) — $9.99
- **Alternatives:** DHT11 (~$2–5, less precise: 0–50 °C ±2 °C, 20–80 %RH; widely sold on Amazon/AliExpress —
  Adafruit's DHT11/DHT22 are discontinued). A 4.7–10 kΩ pull-up on DATA is needed.
- **Driver:** `DHT` supports `DHT_TYPE_DHT11` / `DHT_TYPE_DHT22` (`dht.h`).
- **Wiring:** single-wire DATA pin; 3–5 V power. One wire, careful bit-bang timing.

## Group A · App 23 Ultrasonic Rangefinder (also App 44 Parking Sensor) — HC-SR04

- **Sensor:** **HC-SR04** (driver: `ULTRASONIC`)
- **Buy:** [Adafruit HC-SR04 + 2×10 kΩ resistors](https://www.adafruit.com/product/3942) — $3.95
- **3.3 V-friendly alternative:** [RCWL-1601 (HC-SR04-compatible, 3 V or 5 V)](https://www.adafruit.com/product/4007) — $3.95
- **Driver:** `ULTRASONIC` — `ultrasonic_types.h` constants are HC-SR04 timing (`40 kHz`, echo up to ~8 m).
- **Wiring gotcha:** classic HC-SR04 ECHO is **5 V logic** — the included 10 kΩ divider (or a level shifter)
  protects the 3.3 V GPIO. TRIG can be driven at 3.3 V.

## Group A · App 24 IR Proximity Switch — Sharp GP2Y0A21YK0F

- **Sensor:** **GP2Y0A21YK0F** (driver: `IR_DISTANCE`)
- **Buy:** [Pololu GP2Y0A21YK0F (10–80 cm)](https://www.pololu.com/product/136) — $12.95 ·
  [Adafruit, with cable](https://www.adafruit.com/product/164) — $14.95
- **Alternatives the driver supports** (`ir_distance_types.h`):
  - [GP2Y0A41SK0F — 4–30 cm](https://www.pololu.com/product/2464)
  - [GP2Y0A02YK0F — 20–150 cm](https://www.pololu.com/product/1137)
  - GP2Y0A51SK — 2–15 cm
- **Wiring:** analog output → ADC (5 V power); add a ≥10 µF cap across Vcc/GND near the sensor.
- **Driver:** `IR_DISTANCE` maps ADC counts to distance with per-part min/max constants.

## Group A · App 24 IR Proximity Switch — IR remote pair (TSOP38238 + 940 nm LED)

- **Sensor:** **TSOP38238** receiver + **940 nm IR LED** (driver: `IR`)
- **Buy:** [TSOP38238 38 kHz receiver](https://www.adafruit.com/product/157) — $1.95 ·
  [5 mm 940 nm IR LED](https://www.adafruit.com/product/387) — $0.75
- **Driver:** `IR` transmits/receives **NEC protocol at 38 kHz** (`IR_CARRIER_FREQUENCY 38000`, NEC header 9 ms/4.5 ms).
- **Wiring:** TSOP38238 — pin 3 → VS (+5 V), pin 2 → GND, pin 1 → data (raw demodulated).
- **Note:** a $5–10 any-brand NEC remote (or the LED + a series resistor on a GPIO) completes the pair.

## Group A · App 25 Laser ToF Distance — VL53L0X

- **Sensor:** **VL53L0X** (driver: `LASER_DISTANCE`)
- **Buy:** [Adafruit VL53L0X](https://www.adafruit.com/product/3317) — $14.95 ·
  [Pololu VL53L0X carrier](https://www.pololu.com/product/2490) — $19.95
- **Longer range (4 m):** [Pololu VL53L1X carrier](https://www.pololu.com/product/3415) — $22.95
- **Driver:** `LASER_DISTANCE` defaults to **VL53L0X** (`LASER_DISTANCE_DEFAULT_I2C_ADDRESS 0x52`); VL53L1X is the second mode.
- **Wiring:** I²C (SDA/SCL) + VIN 2.6–5.5 V (on-board regulator + level shifters).
- **Note:** mm-accurate, independent of target color — but don't cover the laser window with the protective liner.
